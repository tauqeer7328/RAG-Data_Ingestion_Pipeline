```python
import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
sample_text = """ Artificial Intelligence (AI) is transforming industries by enabling machines to perform tasks that typically require human intelligence. These tasks include learning from data, recognizing patterns, understanding language, and making decisions.

Machine Learning (ML), a subset of AI, focuses on building systems that improve their performance as they are exposed to more data. Deep Learning, a further subset of ML, uses neural networks with many layers to process complex patterns in large datasets.

Retrieval-Augmented Generation (RAG) is an advanced AI technique that combines information retrieval with text generation. Instead of relying only on pre-trained knowledge, RAG systems fetch relevant documents from a knowledge base and use them to generate more accurate and context-aware responses.

A typical RAG pipeline consists of several stages: data ingestion, document splitting, embedding generation, vector storage, and query retrieval. First, documents are collected and preprocessed. Then, they are split into smaller chunks for efficient processing. Each chunk is converted into embeddings, which are numerical representations of text.

These embeddings are stored in a vector database such as FAISS or Chroma. When a user asks a question, the system retrieves the most relevant chunks based on similarity search. Finally, a language model generates a response using the retrieved context.

RAG systems are widely used in applications like chatbots, document search, customer support automation, and knowledge assistants. They improve accuracy, reduce hallucinations, and allow systems to work with private or domain-specific data."""
with open("../data/text_files/sample.txt", "w",encoding="utf-8") as f:
    f.write(sample_text)
print("Sample text file created at ../data/text_files/sample.txt")
```

```python
#from langchain_core.documents_loaders import TextLoader
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/text_files/sample.txt")
documents = loader.load()
print(documents)
```
